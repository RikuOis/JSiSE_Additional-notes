### 問題
問題：引っ越し荷物を効率良く並べ替えよう
あなたは友人の引っ越しを手伝うことになりました。段ボール箱にはたくさんの荷物が詰まっており、中には本・食器・衣類・書類などさまざまな種類のものが混在しています。そこで効率よく片付けをするために、まずは次の2つの観点から荷物を“ソート（並べ替え）”することを考えました。

大きさ（重ねやすさ）順
大きなものを下に、小さなものを上に積み重ねたい。

壊れやすさ順
壊れやすいものを先に分けておき、丁寧に扱う必要がある。

具体的には、「段ボール箱を開ける→荷物を一時的に取り出す→条件に従って並べ替える→片付ける／運び出す」の流れで作業します。ただし、作業時間をできるだけ短くするために無駄な取り出しや積み直しは減らしたいと考えています。

・大きさ順で並べ替え
最初に大きさに注目して、荷物を大きい順に並べたい（小さいものを上にしたい）。
ただし、同じくらいの大きさの荷物が複数ある場合は、単に見つけた順で後ろに回してもよい（＝“安定ソート”でなくてもよい）。

・壊れやすさ順で並べ替え
食器や割れ物、精密機器など壊れやすい物品をまとめて手前に置き、先に取り扱うようにしたい。
ただし、壊れやすい物品の中でも“精密機器”だけは最優先で扱いたい。もし複数の“精密機器”があったら、箱から出てきた順番を保って並べておきたい（＝“安定ソート”が望ましい場面）。

今回は例として、以下のものを扱うこととする．（品物，大きさ，優先度）
("皿", 2,20), ("衣類", 3,15), ("パソコン", 1,30), ("書類", 4,10), ("ゲーム機",1,25)

質問1.
(大きさ順) 荷物の大きさを大きい順に素早く並べたいとき、どのソートアルゴリズムが考えられますか？その方法でソートする関数を実装しなさい。また、そのアルゴリズムを選ぶ理由やメリット・デメリットを簡単に答えてください。

質問2.
(壊れやすさ順) 壊れやすさに着目して“壊れやすいもの”だけを先に抜き出しつつ、特に“精密機器”を箱から出てきた順のまま最優先で並べたい場合、どのようにソートアルゴリズムを組み合わせるとよいでしょうか？　その方法でソートする関数を実装しなさい。また、“安定ソート”という概念がどのように役に立つか説明してください。


```
### 入出力例
質問１
入力例　[("皿", 30), ("衣類", 20), ("パソコン", 10), ("書類", 15), ("ゲーム機", 25)]
出力例　[('皿', 30), ('ゲーム機', 25), ('衣類', 20), ('書類', 15), ('パソコン', 10)]
質問２
入力例　[("皿", 2), ("衣類", 3), ("パソコン", 1), ("書類", 4), ("ゲーム機", 1)]
出力例　[('パソコン', 1), ('ゲーム機', 1), ('皿', 2), ('衣類', 3), ('書類', 4)]
```



# 解答コード


````python
###解答
#質問1
def quick_sort_by_size(items):
    if len(items) <= 1:
        return items
    pivot = items[len(items) // 2][1]  # 大きさを基準にする
    left = [x for x in items if x[1] > pivot]
    middle = [x for x in items if x[1] == pivot]
    right = [x for x in items if x[1] < pivot]
    return quick_sort_by_size(left) + middle + quick_sort_by_size(right)

def validate_input_for_size(items):
    if not isinstance(items, list):
        raise ValueError("入力はリスト形式で指定してください。")
    for item in items:
        if not (isinstance(item, tuple) and len(item) == 2 and isinstance(item[1], (int, float))):
            raise ValueError("入力リスト内の各要素は ('名前', 数値) の形式で指定してください。")

# 入力例
try:
    items = [("皿", 30), ("衣類", 20), ("パソコン", 10), ("書類", 15), ("ゲーム機", 25)]
    validate_input_for_size(items)  # 入力を検証
    sorted_items = quick_sort_by_size(items)
    print("大きさ順に並べ替えた結果:", sorted_items)
except ValueError as e:
    print("エラー:", e)



#質問2
def merge_sort_by_fragility(items):
    if len(items) <= 1:
        return items
    mid = len(items) // 2
    left = merge_sort_by_fragility(items[:mid])
    right = merge_sort_by_fragility(items[mid:])
    return merge(left, right)

def merge(left, right):
    result = []
    i = j = 0
    while i < len(left) and j < len(right):
        if left[i][1] <= right[j][1]:  # 優先度で比較
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    result.extend(left[i:])
    result.extend(right[j:])
    return result

def validate_input_for_fragility(items):
    if not isinstance(items, list):
        raise ValueError("入力はリスト形式で指定してください。")
    for item in items:
        if not (isinstance(item, tuple) and len(item) == 2 and isinstance(item[1], int)):
            raise ValueError("入力リスト内の各要素は ('名前', 整数) の形式で指定してください。")

# 入力例
try:
    items = [("皿", 2), ("衣類", 3), ("パソコン", 1), ("書類", 4), ("ゲーム機", 1)]
    validate_input_for_fragility(items)  # 入力を検証
    sorted_items = merge_sort_by_fragility(items)
    print("壊れやすさ順に並べ替えた結果:", sorted_items)
except ValueError as e:
    print("エラー:", e)

````

### 解説
問題1
ソートアルゴリズムの選択
ここでは、クイックソート(Quick Sort) を選びます。

理由
クイックソートは平均計算量 𝑂(𝑛log𝑛)で高速です。
安定ソートではありませんが、このタスクでは「同じ大きさの荷物の順番を保つ必要がない」ため、適しています。
メモリ効率が高く、大量のデータにも対応可能です。

メリット: 平均計算量が高速O(nlogn)。リスト内で操作が完結し、メモリ効率が良い。
デメリット: 最悪計算量がO(n^2) になる可能性がある（要素の分布に依存）。

問題2
ソートアルゴリズムの選択
ここでは、マージソート(Merge Sort) を選びます。

理由
安定ソートであり、同じ壊れやすさの物品が出てきた順序を保つことができます。
壊れやすさ（優先度）で分類しつつ、「精密機器（優先度1）」を最優先で扱う必要があるため、安定ソートが不可欠です。

メリット: 安定ソートであるため、同じ優先度の順序を保てる。計算量が最悪でもO(nlogn) で安定。
デメリット: 再帰的に処理を行うため、追加メモリの使用量が増える。
